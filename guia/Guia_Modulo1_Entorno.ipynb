### Paso 3: Visualización de Prueba
Generemos un gráfico sencillo para confirmar que los resultados interactivos funcionan correctamente.

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y, label="Seno(x)", color="blue")
plt.title("Prueba de entorno interactivo - Principios de IA")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.legend()
plt.grid(True)
plt.show()
```
